```python
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestClassifier
import pandas as pd
```

```python
df = pd.DataFrame({
    "cpu": [20,30,40,85,90,95,35,50,88,92],
    "ram": [30,35,40,80,85,90,45,50,82,91],
    "caida": [0,0,0,1,1,1,0,0,1,1]
})
```

```python
X = df[["cpu", "ram"]]
y = df["caida"]
```

```python
parametros = {
    "n_estimators": [50,100],
    "max_depth": [2,4,6]
}

grid = GridSearchCV(
    RandomForestClassifier(),
    parametros,
    cv=5
)

grid.fit(X, y)

print(grid.best_params_)
print(grid.best_score_)
```
