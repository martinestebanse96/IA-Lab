```python
import numpy as np

ventas = np.array([100, 120, 90, 140, 160])

promedio = ventas.mean()
maximo = ventas.max()
aumento = []
for venta in ventas:
    aumento.append(venta * 1.10)
mayores_a_110 = []
for venta in ventas:
    if venta > 110:
        mayores_a_110.append(venta)

print("Promedio:", promedio)
print("Máximo:", maximo)
print("Ventas con aumento del 10%:", aumento)
print("Ventas mayores a 110:", mayores_a_110)

#Resolucion propuesta, np esta optimizado para correr de manera mas eficiente que los loops tradicionales
import numpy as np

ventas = np.array([100,120,90,140,160])

promedio = ventas.mean()
maximo = ventas.max()
aumento = ventas * 1.10
mayores_a_110 = ventas[ventas > 110]

print(promedio)
print(maximo)
print(aumento)
print(mayores_a_110)

```
