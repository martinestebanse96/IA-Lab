```python
numeros = [10, 20, 30, 40, 50]

print(numeros)
print(numeros[0])
print(numeros[-1])
print(len(numeros))

numeros.append(60)
numeros.remove(20)

print(numeros)

for n in numeros:
    print(n)
```

```python
persona = {
    "nombre": "Martin",
    "edad": 30,
    "pais": "Argentina"
}

print(persona["nombre"])
print(persona["pais"])

for clave, valor in persona.items():
    print(clave, valor)
```
