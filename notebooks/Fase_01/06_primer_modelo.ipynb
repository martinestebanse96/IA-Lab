```python
import pandas as pd
from sklearn.linear_model import LinearRegression
```

```python
datos = {
    "m2": [40, 50, 60, 70, 80, 90],
    "precio": [80000, 100000, 120000, 140000, 160000, 180000]
}

df = pd.DataFrame(datos)
```

```python
X = df[["m2"]]      # entrada
y = df["precio"]    # objetivo
```

```python
modelo = LinearRegression()
modelo.fit(X, y)
```

```python
modelo.predict([[75]])
```

```python
modelo.coef_
modelo.intercept_
```

```python
modelo.predict([[55]])
```

```python
#Cambiar dataset y poner ruido realista
datos = {
    "m2": [40, 50, 60, 70, 80, 90, 100],
    "precio": [80000, 100000, 120000, 140000, 160000, 180000, 700000]
}

df = pd.DataFrame(datos)

X = df[["m2"]]      # entrada
y = df["precio"]    # objetivo

modelo = LinearRegression()
modelo.fit(X, y)
```

```python
modelo.predict([[65]])
```

```python
#Ver si sigue funcionando
modelo.predict([[147]])
```
