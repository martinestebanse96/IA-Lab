```python
import pandas as pd
from sklearn.cluster import KMeans
```

```python
df = pd.DataFrame({
    "cpu": [20,25,22,85,88,90,50,55,52],
    "ram": [30,28,35,80,85,82,50,48,52]
})
```

```python
X = df[["cpu","ram"]]

modelo = KMeans(n_clusters=3, random_state=42)
modelo.fit(X)

df["grupo"] = modelo.labels_
print(df)
```

```python
print(modelo.cluster_centers_)
```
